##4.	Operasi matriks dan vektor dengan python

###a.	Membuat matriks di pyhon

```python
import numpy as np

A = np.array([
    [80,30,25],
    [50,20,10]
])

B = np.array([
    [30,60,15],
    [10,30,30]
])

hasil = A + B

print(hasil)
```

### b.	Melakukan operasi dot product

```python
import numpy as np
A = np.array([10, 20, 15])
B = np.array([16, 67, 35])

hasil = np.dot(A, B)

print(hasil)
```

### c.	Melakukan operasi cross product

```python
import numpy as np
A = np.array([10, 20, 15])
B = np.array([16, 67, 35])

Hasil = np.cross(A, B)

print(hasil)
```

## E.	TUGAS PRAKTIKUM

```python
import numpy as np

A = np.array([
    [3, 1],
    [2, 0]
])

B = np.array([
    [2, 1],
    [1, -1]
])

C = np.array([
    [-1, 1, 2],
    [2, 0, -1]
])

hasil_AB = np.matmul(A, B)

print("A =")
print(A)

print("\nB =")
print(B)

print("\nA x B =")
print(hasil_AB)
```

```python
hasil_BC = np.matmul(B, C)

print("B . C =")
print(hasil_BC)
```

```python
 import numpy as np

A = np.array([5,3,7])
B = np.array([12,-3,1])

panjang_A = np.linalg.norm(A)
panjang_B = np.linalg.norm(B)

dot = np.dot(A,B)

cross = np.cross(A,B)

print("Panjang Vektor A =", panjang_A)
print("Panjang Vektor B =", panjang_B)

print("\nA . B =", dot)

print("\nA x B =")
print(cross)
```
